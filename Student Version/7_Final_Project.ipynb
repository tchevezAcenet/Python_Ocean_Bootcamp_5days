Now, let's combine everything we have learned so far to work on a real-life dataset.

This dataset contains information about marine litter found on the seafloor of the southeastern North Sea. It was collected during scientific research aimed at understanding the distribution and impact of marine litter in this area.

- **Citation:** The dataset is attributed to Lars Gutow and published by the Alfred Wegener Institute through PANGAEA. It is part of a larger study of marine litter in the North Sea. You can access it here: https://doi.org/10.1594/PANGAEA.890785

- **Note:** The file you will use is a **teaching version** of this dataset. Extra rows and some data-entry problems were added on purpose so you can practice cleaning data, so your results will not match the published dataset.

- **Location:** The southeastern North Sea.

- **Campaigns:** The samples come from research cruises. The *Event* column tells you which cruise and haul each row comes from (for example, *HE419/004-8* is a haul from cruise HE419 in April 2014, and *HE480/25-2* is a haul from cruise HE480 in 2017).

- **Method/Device:** A beam trawl (BEAM), a net dragged along the seafloor, was used to collect the litter.

## How the data is organized

Each row is **one litter item** found in a trawl haul. Several rows can share the same station and time because more than one item was found in that haul.

## Dataset Columns
- **Event:** The cruise and haul ID (for example, *HE419/004-8*).
- **Station:** The name of the sampling station.
- **Date/Time:** The date and time of sampling at the station.
- **Latitude:** The geographic latitude of the sampling point.
- **Longitude:** The geographic longitude of the sampling point.
- **Elevation [m]:** The position of the seafloor relative to sea level, in meters. Values are negative because the samples are below the surface: *-18* means the seafloor was 18 m deep.
- **Litter obj:** The litter item that was found (for example, *net fiber* or *glass bottle*).
- **Litter cat:** The material category of the item (*plastics*, *metal*, *glass* or *others*).
- **Litter fish:** Whether the item is **fishing-related litter**, such as nets, ropes or fishing line (*yes* or *no*). It does **not** record whether fish were caught.

This dataset helps us understand marine pollution in the North Sea: what kinds of litter end up on the seafloor, where, and at what depths.
